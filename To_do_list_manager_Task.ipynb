# To-Do List Manager

## 📝 Overview
A command-line based To-Do List Manager built with Python that allows users to manage their tasks efficiently. The application features a color-coded interface for better user experience and includes basic task management operations.

## ✨ Features
- **Add Tasks**: Easily add new tasks to your list
- **View Tasks**: Display all tasks with numbered formatting and color coding
- **Remove Tasks**: Remove tasks with confirmation prompt
- **User-Friendly Interface**: Color-coded messages and clear menu options
- **Input Validation**: Robust error handling for invalid inputs

## 🛠️ Technical Details
- **Language**: Python 3.x
- **Data Structure**: Lists for task management
- **UI Enhancement**: ANSI escape codes for colored output
- **Error Handling**: Try-except blocks for input validation

## 💻 Usage
```bash
# Run the program
python todo_list_manager.py

# Menu Options:
1. Add Task    - Add a new task to the list
2. View Tasks  - Display all current tasks
3. Remove Task - Remove a specific task
4. Quit       - Exit the program
```

## 🔍 Code Structure
- `add_task()`: Adds new tasks to the list
- `view_tasks()`: Displays all tasks with formatting
- `remove_task()`: Removes specified tasks with confirmation
- `highlight_task()`: Formats task text with color coding
- `main()`: Controls project is open source and available under the MIT License.


````python
def add_task(tasks, task):
    tasks.append(task)
    print("Task added successfully!")

def view_tasks(tasks):
    print("Your To-Do List:")
    if not tasks:
        print("No tasks yet.")
    else:
        for index, task in enumerate(tasks, start=1):
            print(f"\033[1;32m{index}.{task}\033[0m")

def highlight_task(task):
    # ANSI escape code for bold and red text
    return f"\033[1;31m{task}\033[0m"

def remove_task(tasks):
    view_tasks(tasks)
    if tasks:
        try:
            task_number = int(input("Enter the number of the task to remove: "))
            if 1 <= task_number <= len(tasks):
                task_to_remove = tasks[task_number - 1]
                print(f"Task to remove: {highlight_task(task_to_remove)}")
                confirmation = input("Are you sure you want to remove this task? (y/n): ")
                if confirmation.lower() == 'y':
                    removed_task = tasks.pop(task_number - 1)
                    print(f"Task '{removed_task}' removed successfully!")
                else:
                    print("Task removal canceled.")
            else:
                print("Invalid task number.")
        except ValueError:
            print("Please enter a valid number.")
    else:
        print("No tasks to remove.")

def main():
    tasks = []
    while True:
        print("\n1. Add Task")
        print("2. View Tasks")
        print("3. Remove Task")
        print("4. Quit")
        
        choice = input("Enter your choice: ")
        
        if choice == '1':
            task = input("Enter the task: ")
            add_task(tasks, task)
        elif choice == '2':
            view_tasks(tasks)
        elif choice == '3':
            remove_task(tasks)
        elif choice == '4':
            print("Thank you for using the to-do list manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

main()
````
