# YouTube Video Data Scraper
## Overview
This code demonstrates a simple implementation of the YouTube Data API v3 to search for videos and save their information to a CSV file.

## Dependencies
```python
pip install google-api-python-client
```

## Key Components

### 1. API Setup
```python
from googleapiclient.discovery import build
API_KEY = 'YOUR_API_KEY'
youtube = build('youtube', 'v3', developerKey=API_KEY)
```

### 2. Search Function
```python
def search_videos(query, max_results):
    request = youtube.search().list(
        part='snippet',
        q=query,
        maxResults=max_results
    )
    return request.execute()
```

### 3. Data Extraction
```python
video_data = []
for item in results['items']:
    video_info = {
        'Title': item['snippet']['title'],
        'Channel': item['snippet']['channelTitle'],
        'Published At': item['snippet']['publishedAt'],
        'Description': item['snippet']['description']
    }
    video_data.append(video_info)
```

### 4. CSV Export
```python
csv_columns = ['Title', 'Channel', 'Published At', 'Description']
with open(csv_file, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=csv_columns)
    writer.writeheader()
    for data in video_data:
        writer.writerow(data)
```

````python
video_data = []
for item in results['items']:
    video_info = {
   'Title': item['snippet']['title'],
        'Channel': item['snippet']['channelTitle'],
        'Published At': item['snippet']['publishedAt'],
        'Description': item['snippet']['description']
    }
    video_data.append(video_info)
````

````python
csv_file = 'youtube_videos.csv'
````

````python
csv_columns = ['Title', 'Channel', 'Published At', 'Description']
````

````python
try:
    with open(csv_file, 'w', newline='', encoding='utf-8') as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=csv_columns)
        writer.writeheader()
        for data in video_data:
            writer.writerow(data)
    print(f"Data successfully saved to {csv_file}")
except IOError:
    print("I/O error occurred while saving to CSV")
````
