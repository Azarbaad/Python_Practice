# Web Scraping Quotes Demo 🌐

## Overview
A simple Python script that demonstrates web scraping fundamentals by extracting quotes from quotes.toscrape.com.

## Features
- Fetches inspirational quotes and author information
- Demonstrates basic web scraping concepts
- Shows HTML parsing techniques
- Formats output in a readable way

## Technologies
- **Python** - Core programming language
- **BeautifulSoup4** - HTML parsing library
- **Requests** - HTTP library for making web requests
sts library

```python
import requests
from bs4 import BeautifulSoup

# URL of the website
url = 'http://quotes.toscrape.com'

# Send a GET request to the website
response = requests.get(url)

# Parse the HTML content
soup = BeautifulSoup(response.text, 'html.parser')

# Find all quote elements
quotes = soup.find_all('div', class_='quote')

# Extract and print text and author for each quote
for quote in quotes:
    text = quote.find('span', class_='text').text
    author = quote.find('small', class_='author').text
    print(f'"{text}" - {author}')
```
